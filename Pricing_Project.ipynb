```python
df["Margin"]=df["Revenue"]-df["Cost"]
```

```python
df["Margin_%"]=(df["Margin"]/df["Revenue"])*100
```

```python
df["Discount_Amount"]=df["List_Price"]*(df["Discount_%"]/100)
```

```python
df["Discount_%"]=(df["Revenue"]/df["Discount_Amount"])*100
```

```python
high_discount_df=df[df["Discount_%"]>25]
```

```python
low_margin_df=df[df["Margin_%"]<30]
```

```python
risk_deals=df[(df["Discount_%"]>25)&(df["Margin_%"]<30)]
```

```python
df["Deal_Date"]=pd.to_datetime(df["Deal_Date"])
df.head()
```

```python
df["Month"]=df["Deal_Date"].dt.to_period("M")
```

```python
Monthly_summary=df.groupby("Month").agg({
    "Revenue":"sum",
    "Discount_%":"mean",
    "Margin_%":"mean"
})
print(Monthly_summary)
```

```python
Monthly_summary.to_csv("Monthly_pricing_summary.csv",index=False)
```

```python
#Margin Calculator
def calculate_margin(revenue,cost):
    margin=revenue-cost
    margin_pct=(margin/revenue)*100
    return margin, margin_pct

m,m_pct=calculate_margin(200000,140000)
print(m,m_pct)
```

```python
def risk_flagging(row):
    if (row["Discount_%"]>25) & (row["Margin_%"]<30):
        return "High_Risk"
    else:
        return "Low Risk"

df["Risk_Flag"]=df.apply(risk_flagging,axis=1)
```

```python
monthly_report=df.groupby("Month").agg({
    "Revenue":"sum",
    "Discount_%":"mean",
    "Margin_%":"mean"
})

monthly_report.to_excel("Monthly_price_report.xlsx")

```
